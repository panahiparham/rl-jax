```python
# Import the experiment config, the results loader and the analysis functions.
import matplotlib.pyplot as plt

from analysis.curves import return_curve, subsample
from main import load_results

from config import EXPERIMENT

PLOTS_DIR = EXPERIMENT.results_dir.parent / "plots"
```

```python
# Load DQN's stored Pong runs, if any are stored yet.
results = load_results(EXPERIMENT, "dqn_pong")
```

```python
# Compute each seed's return over time.
curves = {}
if results is not None:
    timesteps, returns = subsample(return_curve(results.reward, results.done))
    curves = dict(zip(results.seed, returns, strict=True))
```

```python
# Style for the per-seed learning-curve plot.
STYLE = {
    "figure.figsize": (9, 6),
    "axes.spines.top": False,
    "axes.spines.right": False,
    "lines.linewidth": 1,
    "legend.frameon": False,
    "legend.loc": "lower right",
}
COLORS = ["tab:blue", "tab:red", "tab:green", "tab:orange"]
LEGEND_LABEL = "DQN seed {seed}"
TITLE = "DQN on Atari Pong"
XLABEL = "Timestep"
YLABEL = "Return"
YLABEL_KW = {"rotation": 0, "ha": "right", "va": "center", "labelpad": 12}
YLIM = (-21, 21)
```

```python
# Plot each seed's return over time as its own line.
figures = {}
if curves:
    with plt.rc_context(STYLE):
        fig, ax = plt.subplots()
        for (seed, curve), color in zip(curves.items(), COLORS, strict=False):
            ax.plot(timesteps, curve, color=color, label=LEGEND_LABEL.format(seed=seed))
        ax.set(title=TITLE, xlabel=XLABEL, ylim=YLIM)
        ax.set_ylabel(YLABEL, **YLABEL_KW)
        ax.legend()
        fig.tight_layout()
    figures["atari_dqn_pong"] = fig
    plt.show()
```

```python
# Save every plotted figure as a PDF in the plots directory.
PLOTS_DIR.mkdir(exist_ok=True)
for name, fig in figures.items():
    fig.savefig(PLOTS_DIR / f"{name}.pdf", bbox_inches="tight")
print(f"saved {len(figures)} plot(s) to {PLOTS_DIR}")
```
